```python
import pandas as pd

# Ensure the 'date' column is in datetime format
total_sales['date'] = pd.to_datetime(total_sales['date'])

# Define the latest date for calculating recency
latest_date = total_sales['date'].max()

# Perform RFM analysis
rfm_analysis = total_sales.groupby('id_customer').agg(
    Recency=('date', lambda x: (latest_date - x.max()).days),  # Recency: Days since last purchase
    Frequency=('date', 'count'),  # Frequency: Number of transactions
    Monetary=('Gross_sales', 'sum'),  # Monetary: Total spending
    Monetary_net=('Net_sales', 'sum'),
    Monetary_PL=('PL_gross_sales','sum'),
    First_Purchase=('date', 'min'),  # First purchase date
    Last_Purchase=('date', 'max')   # Last purchase date
)

# Add inter-purchase metrics for customers with more than one purchase
rfm_analysis['Avg_Interpurchase_Time'] = total_sales.groupby('id_customer')['date'].apply(
    lambda x: (x.sort_values().diff().dt.days.mean() if len(x) > 1 else None)
).values

rfm_analysis['Std_Dev_Interpurchase_Time'] = total_sales.groupby('id_customer')['date'].apply(
    lambda x: (x.sort_values().diff().dt.days.std() if len(x) > 1 else None)
).values

# Add ticket-level metrics
rfm_analysis['Avg_Ticket'] = total_sales.groupby('id_customer')['Gross_sales'].mean().values
rfm_analysis['Std_Dev_Ticket'] = total_sales.groupby('id_customer')['Gross_sales'].std().values

# Reset index to make it a flat DataFrame
rfm_analysis = rfm_analysis.reset_index()

# Display the RFM analysis
print(rfm_analysis)
```

```python
# Adjusted anomaly threshold to 30%
def check_anomaly_threshold(percentile_dict, threshold=0.5):
    values = list(percentile_dict.values())
    diffs = np.diff(values) / np.array(values[:-1])
    return diffs > threshold

# Calculate anomalies
frequency_anomalies = check_anomaly_threshold(percentiles['Frequency'], threshold=0.5)
monetary_anomalies = check_anomaly_threshold(percentiles['Monetary'], threshold=0.5)

# Set thresholds again
if any(frequency_anomalies):
    frequency_threshold = list(percentiles['Frequency'].values())[-1]
else:
    frequency_threshold = max(percentiles['Frequency'].values())  # Fallback: Use the highest frequency percentile

if any(monetary_anomalies):
    monetary_threshold = list(percentiles['Monetary'].values())[-1]
else:
    monetary_threshold = max(percentiles['Monetary'].values())  # Fallback: Use the highest monetary percentile

# Debugging Outputs
print("Frequency Percentiles:", percentiles['Frequency'])
print("Frequency Anomalies Detected:", frequency_anomalies)
print("Frequency Threshold:", frequency_threshold)

print("Monetary Percentiles:", percentiles['Monetary'])
print("Monetary Anomalies Detected:", monetary_anomalies)
print("Monetary Threshold:", monetary_threshold)

# Step 4: Create a new column to mark anomalies
rfm_analysis['Anomaly'] = np.where(
    ((rfm_analysis['Frequency'] > frequency_threshold) & (frequency_threshold is not None)) |
    ((rfm_analysis['Monetary'] > monetary_threshold) & (monetary_threshold is not None)),
    'Anomalous',
    'Normal'
)

# Filter the anomalies into a separate DataFrame
rfm_anomalies = rfm_analysis[rfm_analysis['Anomaly'] == 'Anomalous']

# Remove anomalies from the original RFM analysis
rfm_analysis_filtered = rfm_analysis[rfm_analysis['Anomaly'] == 'Normal']

# Display results
```

```python
# Regularity Analysis for Frequency (F)
rfm_analysis_filtered['Interpurchase_CV'] = rfm_analysis_filtered.apply(
    lambda row: (row['Std_Dev_Interpurchase_Time'] / row['Avg_Interpurchase_Time']
                 if pd.notnull(row['Std_Dev_Interpurchase_Time']) and row['Avg_Interpurchase_Time'] > 0 else None),
    axis=1
)

rfm_analysis_filtered['Regularity_F'] = rfm_analysis_filtered['Interpurchase_CV'].apply(
    lambda x: 'Highly Regular' if pd.notnull(x) and x <= 0.5 else
              'Moderately Regular' if pd.notnull(x) and x <= 1.0 else
              'Moderately Irregular' if pd.notnull(x) and x <= 1.5 else
              'Highly Irregular' if pd.notnull(x) and x > 1.5 else
              'Unknown'
)

# Regularity Analysis for Monetary (M)
rfm_analysis_filtered['Monetary_CV'] = rfm_analysis_filtered.apply(
    lambda row: (row['Std_Dev_Ticket'] / row['Avg_Ticket']
                 if pd.notnull(row['Std_Dev_Ticket']) and row['Avg_Ticket'] > 0 else None),
    axis=1
)

rfm_analysis_filtered['Regularity_M'] = rfm_analysis_filtered['Monetary_CV'].apply(
    lambda x: 'Highly Regular' if pd.notnull(x) and x <= 0.5 else
              'Moderately Regular' if pd.notnull(x) and x <= 1.0 else
              'Moderately Irregular' if pd.notnull(x) and x <= 1.5 else
              'Highly Irregular' if pd.notnull(x) and x > 1.5 else
              'Unknown'
)

# Regularity Analysis for Recency (R)
rfm_analysis_filtered['R_Subjective'] = rfm_analysis_filtered.apply(
    lambda row: 'H' if row['Recency'] <= row['Avg_Interpurchase_Time'] else
                'M' if row['Recency'] <= row['Avg_Interpurchase_Time'] + row['Std_Dev_Interpurchase_Time'] else
                'L' if row['Recency'] > row['Avg_Interpurchase_Time'] + row['Std_Dev_Interpurchase_Time'] else
                'Unknown',
    axis=1
)

# Display updated DataFrame
rfm_analysis_filtered
```
